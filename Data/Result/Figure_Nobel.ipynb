```python
# One notebook cell. Requires pandas and matplotlib.
from pathlib import Path
import textwrap
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

CSV_PATH = Path("/content/drive/MyDrive/temp_semi/nobel_medicine_2026_probabilities.csv")
TOP_N = 5  # Change to 10 or 30 for a fuller ranking.
OUT_DIR = CSV_PATH.resolve().parent

df = pd.read_csv(CSV_PATH, encoding="utf-8-sig").sort_values("rank")
required = {"rank", "plot_label", "discovery_short", "probability_pct_reported"}
if not required.issubset(df.columns):
    raise ValueError(f"Missing CSV columns: {sorted(required - set(df.columns))}")
if len(df) != 30 or not 1 <= TOP_N <= len(df):
    raise ValueError("Expected the 30-row final forecast and TOP_N between 1 and 30.")
if not df["probability_pct_reported"].between(0, 100).all():
    raise ValueError("Reported probabilities must be percentages from 0 to 100.")

shown = df.head(TOP_N).reset_index(drop=True)
# Preserve the PDF's rounded percentages; do not normalize the displayed subset.
values = shown["probability_pct_reported"].to_numpy()
height = 2.7 + 0.85 * TOP_N
ink, muted, accent, secondary = "#172F40", "#556773", "#087F74", "#8AA4B4"

with plt.rc_context({"font.family": "DejaVu Sans", "svg.fonttype": "none"}):
    fig = plt.figure(figsize=(14, height), facecolor="white")
    fig.text(0.065, 1 - 0.34 / height, "KNOWLEDGE LAB + PRESEEN", fontsize=12,
             weight="bold", color=accent, va="top")
    fig.text(0.065, 1 - 0.72 / height, "2026 Nobel Prize in Physiology or Medicine",
             fontsize=25, weight="bold", color=ink, va="top")
    #fig.text(0.065, 1 - 1.18 / height,
    #         f"Top {TOP_N} of 30 exact laureate-set + discovery outcomes",
    #         fontsize=14, color=muted, va="top")

    ax = fig.add_axes([0.515, 1.12 / height, 0.405, (0.85 * TOP_N - 0.05) / height])
    colors = [accent if int(r) == 1 else secondary for r in shown["rank"]]
    ax.barh(range(TOP_N), values, height=0.43, color=colors, zorder=3)
    ax.set_ylim(TOP_N - 0.45, -0.65)
    ax.set_xlim(0, 30)
    ax.set_xticks(range(0, 31, 5))
    ax.xaxis.set_major_formatter(PercentFormatter(xmax=100, decimals=0))
    ax.tick_params(axis="x", length=0, pad=9, labelsize=11, colors=muted)
    ax.set_yticks([])
    ax.grid(axis="x", color="#E4E9EC", linewidth=0.8, zorder=0)
    for spine in ax.spines.values():
        spine.set_visible(False)

    for i, row in shown.iterrows():
        top = int(row["rank"]) == 1
        label = f'{int(row["rank"]):02d}  {row["plot_label"]}'
        ax.text(-1.11, i - 0.12, label, transform=ax.get_yaxis_transform(),
                ha="left", va="center", fontsize=15.5,
                weight="bold" if top else "normal", color=accent if top else ink)
        description = textwrap.fill(str(row["discovery_short"]), width=48)
        ax.text(-1.11, i + 0.19, description, transform=ax.get_yaxis_transform(),
                ha="left", va="center", fontsize=11.5, color=muted)
        ax.text(row["probability_pct_reported"] + 0.5, i,
                f'{row["probability_pct_reported"]:.1f}%',
                ha="left", va="center", fontsize=19 if top else 17,
                weight="bold" if top else "normal", color=accent if top else ink)

    #fig.text(0.065, 0.62 / height,
             #"Conditional on the award matching one of the 30 listed outcomes; Other is excluded.",
             #fontsize=11.5, color=ink, va="center")
    fig.text(0.065, 0.30 / height,
             "Source: Preseen Treatment Final, 4 Oct 2026.",
             fontsize=10.5, color=muted, va="center")
    OUT_DIR.mkdir(parents=True, exist_ok=True)
    stem = OUT_DIR / f"nobel_medicine_2026_top{TOP_N}"
    for suffix in ("png", "svg"):
        fig.savefig(stem.with_suffix(f".{suffix}"), dpi=240, facecolor="white")
    plt.show()
    print(f"Saved: {stem.with_suffix('.png')} and {stem.with_suffix('.svg')}")
```

```python
# One notebook cell. Requires pandas and matplotlib.
from pathlib import Path
import math
import textwrap
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

CSV_PATH = Path("/content/drive/MyDrive/temp_semi/nobel_physics_2026_probabilities.csv")
TOP_N = 30  # 5 / 10 -> single column; 30 -> 1x2 panels (15 per column)
OUT_DIR = CSV_PATH.resolve().parent

df = pd.read_csv(CSV_PATH, encoding="utf-8-sig").sort_values("rank")
required = {"rank", "plot_label", "discovery_short", "probability_pct_reported"}
if not required.issubset(df.columns):
    raise ValueError(f"Missing CSV columns: {sorted(required - set(df.columns))}")
if len(df) != 30 or not 1 <= TOP_N <= len(df):
    raise ValueError("Expected the 30-row final forecast and TOP_N between 1 and 30.")
if not df["probability_pct_reported"].between(0, 100).all():
    raise ValueError("Reported probabilities must be percentages from 0 to 100.")

shown = df.head(TOP_N).reset_index(drop=True)
# Preserve the PDF's rounded percentages; do not normalize the displayed subset.
N_COLS = 2 if TOP_N > 15 else 1
ROWS = math.ceil(TOP_N / N_COLS)          # rows per column
height = 2.7 + 0.85 * ROWS
width = 14 * N_COLS                       # each panel keeps the original 14-inch geometry
pw = 1 / N_COLS                           # panel width as a figure fraction
ink, muted, accent, secondary = "#172F40", "#556773", "#087F74", "#8AA4B4"

with plt.rc_context({"font.family": "DejaVu Sans", "svg.fonttype": "none"}):
    fig = plt.figure(figsize=(width, height), facecolor="white")
    fig.text(0.065 * pw, 1 - 0.34 / height, "KNOWLEDGE LAB + PRESEEN", fontsize=12,
             weight="bold", color=accent, va="top")
    fig.text(0.065 * pw, 1 - 0.72 / height, "2026 Nobel Prize in Physics",
             fontsize=25, weight="bold", color=ink, va="top")

    for c in range(N_COLS):
        part = shown.iloc[c * ROWS:(c + 1) * ROWS].reset_index(drop=True)
        if part.empty:
            break
        values = part["probability_pct_reported"].to_numpy()
        ax = fig.add_axes([(c + 0.515) * pw, 1.12 / height,
                           0.405 * pw, (0.85 * ROWS - 0.05) / height])
        colors = [accent if int(r) == 1 else secondary for r in part["rank"]]
        ax.barh(range(len(part)), values, height=0.43, color=colors, zorder=3)
        ax.set_ylim(ROWS - 0.45, -0.65)   # same row spacing in every column
        ax.set_xlim(0, 30)                # shared scale across columns
        ax.set_xticks(range(0, 31, 5))
        ax.xaxis.set_major_formatter(PercentFormatter(xmax=100, decimals=0))
        ax.tick_params(axis="x", length=0, pad=9, labelsize=11, colors=muted)
        ax.set_yticks([])
        ax.grid(axis="x", color="#E4E9EC", linewidth=0.8, zorder=0)
        for spine in ax.spines.values():
            spine.set_visible(False)

        for i, row in part.iterrows():
            top = int(row["rank"]) == 1
            label = f'{int(row["rank"]):02d}  {row["plot_label"]}'
            ax.text(-1.11, i - 0.12, label, transform=ax.get_yaxis_transform(),
                    ha="left", va="center", fontsize=15.5,
                    weight="bold" if top else "normal", color=accent if top else ink)
            description = textwrap.fill(str(row["discovery_short"]), width=48)
            ax.text(-1.11, i + 0.19, description, transform=ax.get_yaxis_transform(),
                    ha="left", va="center", fontsize=11.5, color=muted)
            ax.text(row["probability_pct_reported"] + 0.5, i,
                    f'{row["probability_pct_reported"]:.1f}%',
                    ha="left", va="center", fontsize=19 if top else 17,
                    weight="bold" if top else "normal", color=accent if top else ink)

    fig.text(0.065 * pw, 0.30 / height,
             "Source: Preseen Treatment Final, 4 Oct 2026.",
             fontsize=10.5, color=muted, va="center")
    OUT_DIR.mkdir(parents=True, exist_ok=True)
    stem = OUT_DIR / f"nobel_physics_2026_top{TOP_N}"
    for suffix in ("png", "svg"):
        fig.savefig(stem.with_suffix(f".{suffix}"), dpi=240, facecolor="white")
    plt.show()
    print(f"Saved: {stem.with_suffix('.png')} and {stem.with_suffix('.svg')}")
```
